**3. Using Comparison Operators**

```r
attendance <- 75
grade <- 70

a_lessthan <- attendance < 75
b_lessthan <- grade < 75
c_greaterthan <- attendance >= 75
d_greaterthan <- grade >= 75

print(a_lessthan)
print(b_lessthan)
print(c_greaterthan)
print(d_greaterthan)
```

**4. Using AND and OR**

```r
if (attendance < 75 & grade < 75){
  print("you are at risk")
}
if (attendance < 75 | grade < 75){
  print("you are at risk")
}
```
