**3.Numerical Operations**

```r
x <- 10
y <- 2

add <- x + y #Addition
sub <- x - y #Subtraction
mul <- x * y #Multiplication
div <- x / y #Division
mod <- x %% y #Modulus
exp <- x ^ y #Exponent

print(add)
print(sub)
print(mul)
print(div)
print(mod)
print(exp)
```

**8.Vectors**


```r
grades <- c(99,80,98,97,95)

grades

cat("MEAN: ", mean(grades))
cat("\nMAX: ",max(grades))
cat("\nMIN: ",min(grades))
cat("\nLENGTH: ",length(grades))
cat("\nMEDIAN: ",median(grades))
summary(grades)
```
